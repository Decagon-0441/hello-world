```python
#this code will do something

word = input("Enter a name")
num = input("Enter a reasonable integer")

def spacer(name, spaces):
    output = ""
    for i in range(len(name)):
        output += name[i]
        if name[i] != ' ':
            for k in range(int(spaces)):
                output += "+"
    print(output)

spacer(word, num)
```
